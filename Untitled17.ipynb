### Step 1: Import Libraries and Load Data
First, we will import the necessary libraries (pandas, numpy, matplotlib, and seaborn) and load the dataset to examine its basic structure and initial rows.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("air_quality_dataset.csv")

# Display basic information and the first 5 rows of the dataset
print("--- Dataset Information ---")
print(df.info())
print("\n--- First 5 Rows ---")
display(df.head())
```

### Step 2: Data Cleaning (Handling Missing Values)
In this step, we identify if there are any missing (null) values in the dataset and handle them appropriately by imputation or removal.

```python
# Check for missing values in each column
missing_values = df.isnull().sum()
print("Missing values per column:\n", missing_values)

# Fill missing values in numeric columns with their respective mean
numeric_cols = df.select_dtypes(include=[np.number]).columns
for col in numeric_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mean())

# Confirm that there are no remaining missing values
print("\nMissing values after cleaning:", df.isnull().sum().sum())
```

### Step 3: Exploratory Data Analysis (EDA) & Summary Statistics
We will now generate statistical summaries of our dataset and compute the correlation matrix to understand the relationships between different variables.

```python
print("--- Statistical Summary ---")
display(df.describe())

# Visualize the correlation matrix using a heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Air Quality Features')
plt.show()
```

### Step 4: Data Visualization (Trends & Distributions)
In this step, we visualize the distribution and detect outliers in our primary target feature (such as AQI or PM2.5).

```python
# Automatically identify a target column (e.g., AQI or PM2.5) if it exists in the dataset
target_column = None
possible_targets = ['AQI', 'aqi', 'PM2.5', 'pm25', 'Air Quality']
for col in possible_targets:
    if col in df.columns:
        target_column = col
        break

if target_column:
    plt.figure(figsize=(12, 5))

    # Plot distribution
    plt.subplot(1, 2, 1)
    sns.histplot(df[target_column], kde=True, color='skyblue')
    plt.title(f'Distribution of {target_column}')

    # Plot boxplot for outlier detection
    plt.subplot(1, 2, 2)
    sns.boxplot(y=df[target_column], color='salmon')
    plt.title(f'Boxplot of {target_column}')

    plt.tight_layout()
    plt.show()
else:
    # Fallback to the first available numeric column if no standard target is matched
    first_num_col = numeric_cols[0] if len(numeric_cols) > 0 else None
    if first_num_col:
        plt.figure(figsize=(6, 4))
        sns.histplot(df[first_num_col], kde=True)
        plt.title(f'Distribution of {first_num_col}')
        plt.show()
```

### Step 6: Feature Engineering (Categorizing Air Quality)
Let's create a new categorical column `Air_Quality_Class` based on standard PM2.5 thresholds to help classify the air quality of each record.

```python
# Define a function to classify PM2.5 concentration levels
def classify_air_quality(pm25_val):
    if pm25_val <= 12.0:
        return 'Good'
    elif pm25_val <= 35.4:
        return 'Moderate'
    elif pm25_val <= 55.4:
        return 'Unhealthy for Sensitive Groups'
    else:
        return 'Poor/Unhealthy'

# Apply the classification to our dataframe
df['Air_Quality_Class'] = df['PM2.5'].apply(classify_air_quality)

# Count the frequency of each class
class_counts = df['Air_Quality_Class'].value_value_counts() if hasattr(df['Air_Quality_Class'], 'value_value_counts') else df['Air_Quality_Class'].value_counts()
print("Air Quality Class Distribution:")
print(class_counts)
```

### Step 7: Advanced Data Visualizations
Let's plot the distribution of all individual pollutant columns and visualize the relationship of other variables against PM2.5 using regression plots.

```python
# Plot distributions for all pollutants
pollutants = ['PM10', 'NOx', 'NO2', 'SO2', 'VOCs', 'CO', 'CO2', 'CH4']

plt.figure(figsize=(15, 10))
for i, col in enumerate(pollutants, 1):
    plt.subplot(3, 3, i)
    sns.histplot(df[col], kde=True, color='green')
    plt.title(f'{col} Distribution')

plt.tight_layout()
plt.show()

# Regplot of Temperature vs PM2.5
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x='Temperature', y='PM2.5', scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title('Relationship between Temperature and PM2.5')
plt.show()
```
